```python
def is_bad(review_status, rating):
    if review_status == "BAD":
        return True
    elif rating is None :
        return True
    else:
        return False


print(is_bad("OK",4))
print(is_bad("BAD",5))
print(is_bad("OK",None))
    
```

```python
def sum_ignore_none(values):
    nums = 0
    for i in values:
        if i is None :
            continue
        nums += i
    return nums

print(sum_ignore_none([1, None, 5]))
```
