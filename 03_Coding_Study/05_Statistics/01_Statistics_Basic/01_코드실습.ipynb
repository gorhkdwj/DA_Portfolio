```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. 데이터 생성 (Observed Data)
data = {
    'Action': [70, 30],   # Male, Female
    'Romance': [30, 70]   # Male, Female
}
df = pd.DataFrame(data, index=['Male', 'Female'])

print("========== [1. 관측값 (Observed Data)] ==========")
print(df)
print("="*50 + "\n")

# 2. 카이제곱 검정 및 기대값 계산
chi2, p, dof, expected = stats.chi2_contingency(df)
df_expected = pd.DataFrame(expected, columns=df.columns, index=df.index)

print(f"========== [2. 카이제곱 검정 결과] ==========")
print(f"Chi-square Statistic : {chi2:.4f}")
print(f"p-value              : {p:.4f}")
print(f"Degrees of Freedom   : {dof}")
print("\n>>> [기대값 데이터프레임 (Expected Data)]")
print(df_expected)
print("="*50 + "\n")

# 3. 데이터 구조 변환 (Melt)
obs_melted = df.reset_index().melt(id_vars='index', var_name='Genre', value_name='Count')
obs_melted.columns = ['Gender', 'Genre', 'Count']
obs_melted['Type'] = 'Observed'

print("========== [3-1. 관측값 Melt 결과] ==========")
print(obs_melted)
print("-" * 30)

exp_melted = df_expected.reset_index().melt(id_vars='index', var_name='Genre', value_name='Count')
exp_melted.columns = ['Gender', 'Genre', 'Count']
exp_melted['Type'] = 'Expected'

print("========== [3-2. 기대값 Melt 결과] ==========")
print(exp_melted)
print("="*50 + "\n")

plot_data = pd.concat([obs_melted, exp_melted])

print("========== [4. 시각화를 위한 최종 통합 데이터] ==========")
print(plot_data.sort_values(by=['Gender', 'Genre']))
print("="*50 + "\n")

# 4. 시각화
g = sns.catplot(
    data=plot_data, 
    kind='bar', 
    x='Genre', 
    y='Count', 
    hue='Type', 
    col='Gender', 
    palette='muted',
    height=4, 
    aspect=1
)

# 그래프 위에 수치 표시 (변수 이름을 patch로 변경)
for ax in g.axes.flat:
    for patch in ax.patches:  # p -> patch로 변경
        height = patch.get_height()
        if height > 0:
            ax.annotate(f'{height:.1f}', 
                        (patch.get_x() + patch.get_width() / 2., height), 
                        ha='center', va='bottom', 
                        fontsize=10, color='black',
                        xytext=(0, 2), 
                        textcoords='offset points')

# 제목 추가
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle(f'Observed vs Expected (p-value: {p:.4f})')

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# ========================================
# 1. 가상의 실험 데이터 생성
# ========================================
# 가정: 총 400개의 완두콩을 관찰했다고 가정
total = 400

# 관측값 (Observed): 실제 실험에서 관찰된 값 (가상)
observed_yellow = 315  # 노란색 완두
observed_green = 85    # 녹색 완두
observed = np.array([observed_yellow, observed_green])

print("========== [1. 관측값 (Observed)] ==========")
print(f"노란색 완두: {observed_yellow}개")
print(f"녹색 완두  : {observed_green}개")
print(f"총합      : {total}개")
print("="*50 + "\n")

# ========================================
# 2. 멘델의 법칙에 따른 기대값 계산
# ========================================
# 이론적 비율: 노란색(우성) : 녹색(열성) = 3 : 1
ratio_yellow = 3/4
ratio_green = 1/4

expected_yellow = total * ratio_yellow  # 400 * 0.75 = 300
expected_green = total * ratio_green    # 400 * 0.25 = 100
expected = np.array([expected_yellow, expected_green])

print("========== [2. 기대값 (Expected by 3:1 Ratio)] ==========")
print(f"기대 노란색: {expected_yellow}개 (75%)")
print(f"기대 녹색  : {expected_green}개 (25%)")
print("="*50 + "\n")

# ========================================
# 3. 카이제곱 적합도 검정 (Goodness-of-Fit Test)
# ========================================
# scipy.stats.chisquare: 관측값과 기대값 비교
chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)

print("========== [3. 카이제곱 적합도 검정 결과] ==========")
print(f"Chi-square Statistic: {chi2_stat:.4f}")
print(f"p-value             : {p_value:.4f}")
print(f"자유도 (df)         : {len(observed) - 1}")
print("-"*50)

alpha = 0.05
if p_value > alpha:
    print(f"결론: p-value({p_value:.4f}) > α({alpha})")
    print("      → 귀무가설 채택: 관측값은 3:1 비율을 따른다고 볼 수 있음")
else:
    print(f"결론: p-value({p_value:.4f}) ≤ α({alpha})")
    print("      → 귀무가설 기각: 관측값은 3:1 비율을 따르지 않음")
print("="*50 + "\n")

# ========================================
# 4. 시각화
# ========================================
categories = ['노란색 (Yellow)', '녹색 (Green)']
x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))

# 막대 그래프
bars_obs = ax.bar(x - width/2, observed, width, label='관측값 (Observed)', color='gold', edgecolor='black')
bars_exp = ax.bar(x + width/2, expected, width, label='기대값 (Expected)', color='lightgreen', edgecolor='black')

# 막대 위에 수치 표시
for bar in bars_obs:
    height = bar.get_height()
    ax.annotate(f'{int(height)}', 
                xy=(bar.get_x() + bar.get_width()/2, height),
                ha='center', va='bottom', fontsize=12, fontweight='bold')

for bar in bars_exp:
    height = bar.get_height()
    ax.annotate(f'{int(height)}', 
                xy=(bar.get_x() + bar.get_width()/2, height),
                ha='center', va='bottom', fontsize=12, fontweight='bold')

# 그래프 꾸미기
ax.set_ylabel('빈도 (Frequency)', fontsize=12)
ax.set_xlabel('완두콩 색상', fontsize=12)
ax.set_title(f"멘델의 법칙 검정 (3:1 비율)\nχ² = {chi2_stat:.2f}, p-value = {p_value:.4f}", fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11)
ax.legend()
ax.set_ylim(0, max(observed.max(), expected.max()) * 1.15)

plt.tight_layout()
plt.show()

# ========================================
# 5. 데이터프레임으로 요약
# ========================================
summary_df = pd.DataFrame({
    '색상': categories,
    '관측값': observed,
    '기대값': expected,
    '(O-E)²/E': (observed - expected)**2 / expected
})
print("========== [5. 요약 테이블] ==========")
print(summary_df)
print(f"\n카이제곱 통계량 = Σ(O-E)²/E = {sum((observed - expected)**2 / expected):.4f}")
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# ========================================
# 1. 불균등한 관측값 데이터 생성
# ========================================
# 상황: 설문조사에서 남성 150명, 여성 50명 응답 (불균등)
# 장르: Action을 좋아하는 사람이 더 많음 (불균등)

observed = np.array([
    [90, 60],   # Male:   Action 90, Romance 60  (합계: 150)
    [30, 20]    # Female: Action 30, Romance 20  (합계: 50)
])

# 데이터프레임으로 시각화
df_obs = pd.DataFrame(
    observed, 
    index=['Male', 'Female'], 
    columns=['Action', 'Romance']
)

# 행/열 합계 추가
df_obs['행 합계'] = df_obs.sum(axis=1)
df_obs.loc['열 합계'] = df_obs.sum(axis=0)

print("========== [1. 관측값 (Observed)] ==========")
print(df_obs)
print(f"\n전체 합계: {observed.sum()}")
print("="*50 + "\n")

# ========================================
# 2. 카이제곱 독립성 검정 - 기대값 자동 계산
# ========================================
chi2, p, dof, expected = stats.chi2_contingency(observed)

df_exp = pd.DataFrame(
    expected, 
    index=['Male', 'Female'], 
    columns=['Action', 'Romance']
)

# 행/열 합계 추가
df_exp['행 합계'] = df_exp.sum(axis=1)
df_exp.loc['열 합계'] = df_exp.sum(axis=0)

print("========== [2. 기대값 (Expected - 자동 계산)] ==========")
print(df_exp)
print("="*50 + "\n")

# ========================================
# 3. 기대값 계산 과정 상세 설명
# ========================================
print("========== [3. 기대값 계산 과정] ==========")
total = observed.sum()
row_sums = observed.sum(axis=1)  # [150, 50]
col_sums = observed.sum(axis=0)  # [120, 80]

print(f"행 합계: Male={row_sums[0]}, Female={row_sums[1]}")
print(f"열 합계: Action={col_sums[0]}, Romance={col_sums[1]}")
print(f"전체 합계: {total}\n")

print("계산식: E = (행 합계 × 열 합계) / 전체 합계")
print("-"*50)
print(f"Male-Action   = ({row_sums[0]} × {col_sums[0]}) / {total} = {row_sums[0]*col_sums[0]/total}")
print(f"Male-Romance  = ({row_sums[0]} × {col_sums[1]}) / {total} = {row_sums[0]*col_sums[1]/total}")
print(f"Female-Action = ({row_sums[1]} × {col_sums[0]}) / {total} = {row_sums[1]*col_sums[0]/total}")
print(f"Female-Romance= ({row_sums[1]} × {col_sums[1]}) / {total} = {row_sums[1]*col_sums[1]/total}")
print("="*50 + "\n")

# ========================================
# 4. 검정 결과
# ========================================
print("========== [4. 검정 결과] ==========")
print(f"Chi-square: {chi2:.4f}")
print(f"p-value   : {p:.4f}")
print(f"자유도    : {dof}")
print("="*50)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
from itertools import combinations

# ========================================
# 데이터 생성: 성별, 연령대, 장르 선호
# ========================================
np.random.seed(42)
n = 300

data = pd.DataFrame({
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.choice(['20대', '30대', '40대'], n),
    'Genre': np.random.choice(['Action', 'Romance', 'Comedy'], n)
})

print("========== [데이터 미리보기] ==========")
print(data.head(10))
print(f"\n전체 데이터: {len(data)}명")
print("="*50 + "\n")

# ========================================
# 모든 변수 쌍에 대해 독립성 검정 수행
# ========================================
variables = ['Gender', 'Age', 'Genre']

print("========== [쌍별 독립성 검정 결과] ==========")
for var1, var2 in combinations(variables, 2):
    # 분할표 생성
    contingency = pd.crosstab(data[var1], data[var2])
    
    # 카이제곱 검정
    chi2, p, dof, expected = stats.chi2_contingency(contingency)
    
    print(f"\n[{var1} × {var2}]")
    print(contingency)
    print(f"Chi-square: {chi2:.4f}, p-value: {p:.4f}")
    
    if p < 0.05:
        print("→ 관련 있음 (독립 아님)")
    else:
        print("→ 독립 (관련 없음)")
    print("-"*40)
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import glm
import pandas as pd
import numpy as np

# 데이터 생성
np.random.seed(42)
n = 300

data = pd.DataFrame({
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.choice(['20대', '30대'], n),
    'Genre': np.random.choice(['Action', 'Romance'], n)
})

# 빈도표 생성
freq_table = data.groupby(['Gender', 'Age', 'Genre']).size().reset_index(name='Count')

print("========== [빈도표] ==========")
print(freq_table)

# 로그-선형 모델 적합 (포아송 회귀)
# 완전 독립 모델 vs 상호작용 포함 모델 비교 가능
model = glm('Count ~ Gender + Age + Genre', data=freq_table, family=sm.families.Poisson()).fit()

print("\n========== [로그-선형 모델 결과] ==========")
print(model.summary())
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import glm
import pandas as pd
import numpy as np

# 데이터 생성
np.random.seed(42)
n = 300

data = pd.DataFrame({
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.choice(['20대', '30대'], n),
    'Genre': np.random.choice(['Action', 'Romance'], n)
})

# 빈도표 생성
freq_table = data.groupby(['Gender', 'Age', 'Genre']).size().reset_index(name='Count')

print("========== [빈도표] ==========")
print(freq_table)

# 로그-선형 모델 적합 (포아송 회귀)
# 완전 독립 모델 vs 상호작용 포함 모델 비교 가능
model = glm('Count ~ Gender + Age + Genre', data=freq_table, family=sm.families.Poisson()).fit()

print("\n========== [로그-선형 모델 결과] ==========")
print(model.summary())
```
