### 연관 검색어 수집 기본 API TEST

```python
import requests

service_url = 'https://ac.search.naver.com/nx/ac'

search_query = '인공지능'
params = {
    'q' : search_query,
    'con' : 0,
    'frm' : 'nv',
    'ans' : 2,
    'r_format' : 'json',
    'r_enc' : 'UTF-8',
    'st' : 100
}

response = requests.get(service_url, params=params)

if response.status_code == 200:
    print(response.json())
    result = response.json()
    items = result['items'][0]
    print(items)
```

```python
# 재사용성을 위한 함수화
def get_related_searches(search_query):
    service_url = 'https://ac.search.naver.com/nx/ac'

    params = {
        'q' : search_query,
        'con' : 0,
        'frm' : 'nv',
        'ans' : 2,
        'r_format' : 'json',
        'r_enc' : 'UTF-8',
        'st' : 100
    }

    response = requests.get(service_url, params=params)

    if response.status_code == 200:
        result = response.json()
        items = result['items'][0]
        return(items)


get_related_searches('강아지')
```
