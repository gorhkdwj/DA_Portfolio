### 네이버 증권 삼성전자 일별 시세 추출

https://finance.naver.com/item/sise_day.naver?code=005930&page=1

```python
import requests
import pandas as pd
import time

service_url = 'https://finance.naver.com/item/sise_day.naver'

code = '005930'

params = {
    'code' : code
}

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

for i in range(1,11):
    params['page'] = i
    response = requests.get(service_url, params=params, headers=headers)

    if response.status_code == 200:
        print(f"--- {i}페이지 HTML 정보 출력 ---")
        print(response.text)
        print("-" * 50)
```

```python
import requests
import pandas as pd
import time
from io import StringIO

# 1. 대상 URL 및 기본 설정
service_url = 'https://finance.naver.com/item/sise_day.naver'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

# 모든 페이지의 데이터를 담을 리스트
all_tables = []

# 2. 1~10페이지 루프 진행
for i in range(1, 101):
    params = {'code': '005930', 'page': i}
    response = requests.get(service_url, params=params, headers=headers)
    
    # pandas의 read_html을 사용해 HTML 내의 테이블(표)을 데이터프레임으로 읽기
    # [0]을 붙여 리스트의 첫 번째 테이블을 가져옵니다.
    df = pd.read_html(StringIO(response.text), encoding='euc-kr')[0]
    
    # 리스트에 추가
    all_tables.append(df)
    
    print(f"{i}페이지 수집 완료")

# 3. 데이터 통합 및 정제
# 리스트에 담긴 10개의 데이터프레임을 하나로 합칩니다.
full_df = pd.concat(all_tables, ignore_index=True)

# 네이버 시세 표 중간중간에 있는 빈 줄(NaN)을 제거합니다.
full_df.dropna(inplace=True)

# 인덱스(번호)를 0부터 차례대로 다시 매깁니다.
full_df.reset_index(drop=True, inplace=True)

# 결과 확인
print("\n--- 최종 수집된 데이터 (상위 10행) ---")
full_df
```
