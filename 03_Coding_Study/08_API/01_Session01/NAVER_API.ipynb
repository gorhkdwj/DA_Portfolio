### 블로그 검색

```python
client_id = 'EZMe3ZxCpb5f3YwItH9V'
client_secret = 'qNTnbGijHl'
```

```python
import requests
from pprint import pprint

# 검색어 설정
search_query = '인공지능'

# API 요청 URL
service_url = 'https://openapi.naver.com/v1/search/blog.json'
headers = {
    'X-Naver-Client-Id' : client_id,
    'X-Naver-Client-Secret' : client_secret
}

# API 요청 파라미터 설정
params = {
    'query' : search_query,
    'display' : 20,
    'start' : 1,
    'sort' : 'date'
}

# HTTP GET 요청하기
response = requests.get(
    service_url,
    headers=headers,
    params=params
)
# # HTTP POST로 요청하고 싶다면?
# response = requests.post(
#     service_url,
#     headers=headers,
#     params=params
# )

print(response.status_code) # 응답코드

blog_results =[]

if response.status_code ==200:
    result = response.json() # json 형태의 응답 텍스트 결과를 딕셔너리로 변환
    pprint(result)
    for item in result['items']:
        blog = {
            '제목': item['title'],
            'link' : item['link'],
            'description' : item['description'],
            'postdate' : item['postdate'],
        }
        print(blog)
        blog_results.append(blog)
        
```

```python
len(blog_results)
import pandas as pd
df = pd.DataFrame(blog_results)
df.to_csv(f'블로그 검색_{search_query}.csv', index=False)
```

### 이미지 검색

```python
import requests
from pprint import pprint

# 검색어 설정
search_query = '고양이'

# API 요청 URL
service_url = 'https://openapi.naver.com/v1/search/image'
headers = {
    'X-Naver-Client-Id' : client_id,
    'X-Naver-Client-Secret' : client_secret
}

# API 요청 파라미터 설정
params = {
    'query' : search_query, # UTF-8 인코딩을 해야된다고 적혀있는데, requests 라이브러리는 자동으로 해줌!
    'display' : 20,
    'start' : 2,
    'sort' : 'sim',
    'filter' : 'small'
}

# HTTP GET 요청하기
response = requests.get(
    service_url,
    headers=headers,
    params=params
)
# # HTTP POST로 요청하고 싶다면?
# response = requests.post(
#     service_url,
#     headers=headers,
#     params=params
# )

print(response.status_code) # 응답코드

image_results =[]

if response.status_code ==200:
    result = response.json() # json 형태의 응답 텍스트 결과를 딕셔너리로 변환
    pprint(result)
    for item in result['items']:
        image = {
            '제목': item['title'],
            'link' : item['link'],
            'sizeheight' : item['sizeheight'],
            'sizewidth' : item['sizewidth'],
        }
        print(image)
        image_results.append(image)
        
```

```python
len(image_results)
import pandas as pd
df = pd.DataFrame(image_results)
df.to_csv(f'블로그 검색_{search_query}.csv', index=False)
```

### 여러 페이지의 이미지 가져오기

```python
import requests
from pprint import pprint

# 검색어 설정
search_query = '강아지'

# API 요청 URL
service_url = 'https://openapi.naver.com/v1/search/image'
headers = {
    'X-Naver-Client-Id' : client_id,
    'X-Naver-Client-Secret' : client_secret
}

# 20개씩 5번을 가져와서 총 100개의 이미지를 수집!!
image_results =[]
for i in range(5) : 

# API 요청 파라미터 설정
    params = {
        'query' : search_query, # UTF-8 인코딩을 해야된다고 적혀있는데, requests 라이브러리는 자동으로 해줌!
        'display' : 100,
        'start' : 1+i*100,
        'sort' : 'sim',
        'filter' : 'small'
    }

# HTTP GET 요청하기
    response = requests.get(
        service_url,
        headers=headers,
        params=params
    )

    if response.status_code ==200:
        result = response.json() # json 형태의 응답 텍스트 결과를 딕셔너리로 변환
        pprint(result)
        for item in result['items']:
            image = {
                '제목': item['title'],
                'link' : item['link'],
                'sizeheight' : item['sizeheight'],
                'sizewidth' : item['sizewidth'],
            }
            print(image)
            image_results.append(image)
        
```

```python
len(image_results)
import pandas as pd
df = pd.DataFrame(image_results)
df.to_csv(f'블로그 검색_{search_query}.csv', index=False)
```
