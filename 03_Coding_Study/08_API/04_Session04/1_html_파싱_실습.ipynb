#### 핵심 개념

**`soup` 객체란?**
- `BeautifulSoup(html, 'html.parser')`로 생성한 객체
- HTML 문서 전체를 파싱한 결과물
- HTML 문서의 **최상위(루트)** 요소를 나타냄
- 모든 HTML 요소를 탐색하고 검색하는 시작점
```python
soup = BeautifulSoup(html, 'html.parser')  # soup 객체 생성
```

**`element` (요소)란?**
- HTML의 개별 태그를 나타내는 객체 (예: `<div>`, `<p>`, `<a>` 등)
- `soup.find()` 또는 `soup.select_one()` 등으로 찾은 결과
- `Tag` 타입의 객체로, 해당 태그의 속성, 텍스트, 하위 요소 등에 접근 가능
```python
# element 예시
element = soup.find('div')  # div 태그를 찾아서 element에 저장
print(type(element))  # <class 'bs4.element.Tag'>
print(element.text)   # element의 텍스트 추출
print(element['class'])  # element의 속성 접근
```

**관계 이해:**
```plain
soup (BeautifulSoup 객체)
  └─ html (element)
      └─ body (element)
          ├─ div (element)
          │   └─ p (element)
          └─ a (element)
```

#### 주요 객체 및 함수

| 분류 | 메서드/속성 | 반환 타입 | 설명 |
|------|------------|-----------|------|
| **단일 검색** | `find()` | 요소(Tag) or `None` | 첫 번째 일치 요소 반환 |
| | `select_one()` | 요소(Tag) or `None` | CSS 선택자로 첫 번째 요소 반환 |
| **다중 검색** | `find_all()` | 리스트 `[Tag, ...]` | 모든 일치 요소를 리스트로 반환 |
| | `select()` | 리스트 `[Tag, ...]` | CSS 선택자로 모든 요소를 리스트로 반환 |
| **텍스트 추출** | `.text` | 문자열 | 요소의 모든 텍스트 반환 |
| | `.get_text()` | 문자열 | 옵션 지정 가능한 텍스트 반환 |
| **속성 접근** | `['속성명']` | 값 or 에러 | 속성값 직접 접근 (없으면 에러) |
| | `.get('속성명')` | 값 or `None` | 속성값 안전하게 반환 |
| **관계 탐색** | `.parent` | 요소(Tag) or `None` | 부모 요소 |
| | `.children` | 반복자 | 직계 자식 요소들 |
| | `.next_sibling` | 요소(Tag) or `None` | 다음 형제 요소 |

### BeautifulSoup의 주요 메서드

1. **기본 요소 찾기**
```python
# 단일 요소 찾기
soup.find('태그명')  # 첫 번째 일치하는 태그 찾기
soup.find(id='아이디명')  # id로 찾기
soup.find(class_='클래스명')  # class로 찾기

# 여러 요소 찾기
soup.find_all('태그명')  # 모든 일치하는 태그 찾기
soup.find_all(['태그1', '태그2'])  # 여러 태그 동시에 찾기
soup.find_all('태그명', limit=2)  # 개수 제한하여 찾기
```

2. **CSS 선택자**
```python
soup.select('CSS선택자')  # CSS 선택자로 모든 요소 찾기
soup.select_one('CSS선택자')  # CSS 선택자로 첫 요소 찾기

# 예시
soup.select('.클래스명')  # 클래스로 찾기
soup.select('#아이디명')  # id로 찾기
soup.select('div p')  # div 안의 모든 p 태그 찾기
```

3. **요소 탐색**
```python
# 상하 관계
element.parent  # 부모 요소
element.parents  # 모든 상위 요소들
element.children  # 직계 자식 요소들
element.descendants  # 모든 하위 요소들

# 형제 관계
element.next_sibling  # 다음 형제
element.previous_sibling  # 이전 형제
element.find_next_siblings()  # 다음 형제들 모두
element.find_previous_siblings()  # 이전 형제들 모두
```

4. **콘텐츠 추출**
```python
element.text  # 텍스트 내용 추출
element.get_text()  # 텍스트 내용 추출 (옵션 지정 가능)
```

5. **속성 다루기**
```python
element['속성명']  # 속성값 직접 접근
element.get('속성명')  # 속성값 안전하게 가져오기
element.get('속성명', '기본값')  # 없으면 기본값 반환
element.attrs  # 모든 속성을 딕셔너리로 가져오기
```

6. **실전에서 자주 쓰는 조합**
```python
# 특정 조건의 요소 찾기
soup.find('div', class_='container')
soup.find('태그', attrs={'id': 'ID', 'class': 'CLASS'})

# 여러 단계 탐색
soup.find('div').find('p').find('span')
```

---

주의사항:
1. 요소가 없을 수 있으므로 None 체크 필요
```python
   element = soup.find('div', class_='not-exist')
   if element:  # None 체크
       print(element.text)
```

2. 텍스트 추출 시 공백 처리 고려
```python
   text = element.text.strip()  # 앞뒤 공백 제거
   text = element.get_text(strip=True)  # 공백 제거 옵션
```

3. 복잡한 구조는 단계별 접근이 안전
```python
   # 한 번에 접근 (위험)
   result = soup.find('div').find('ul').find('li').text
   
   # 단계별 접근 (안전)
   div = soup.find('div')
   if div:
       ul = div.find('ul')
       if ul:
           li = ul.find('li')
           if li:
               result = li.text
```

4. 속성 접근 시 get() 메서드 권장
```python
   # 직접 접근 (속성이 없으면 에러)
   href = link['href']
   
   # get() 사용 (안전)
   href = link.get('href')  # 없으면 None 반환
   href = link.get('href', '기본값')  # 없으면 기본값 반환
```

````python
# 라이브러리 임포트
from bs4 import BeautifulSoup

# 파일 불러오기
with open('bs4_학습.html', encoding='utf-8') as f:
    sample_html = f.read()
    
# BeautifulSoup 객체 생성
soup = BeautifulSoup(sample_html, 'html.parser')

# html 출력
print(soup.prettify())
````

## 2. 기본적인 요소 찾기

### 2.1 단일 요소 찾기 (find)

````python
# 첫 번째 p 태그 찾기
first_p = soup.find('p')
print(first_p.text.strip())
````

````python
# id로 요소 찾기
title = soup.find(id='title')
print(title.text.strip())  
````

````python
# class로 요소 찾기
container = soup.find(class_='content')
print(container.text.strip())
````

````python
# 특정 태그의 특정 클래스로 요소 찾기
first_li = soup.find('li', class_='second')
print(first_li.text.strip()) 
````

### 2.2 여러 요소 찾기 (find_all)

````python
# 모든 p 태그 찾기
all_p = soup.find_all('p')
print("\n모든 p 태그 내용:")
for p in all_p:
    print("-", p.text.strip())  # 출력: - 첫 번째 문단, - 두 번째 문단
````

````python
# 클래스가 content인 모든 요소 찾기
items = soup.find_all(class_='content')
print("\n리스트 항목:")
for item in items:
    print("-", item.text.strip())  
````

````python
# li태그에서 클래스가 item 모든 요소 찾기
items = soup.find_all('li', class_='item')
print("\n리스트 항목:")
for item in items:
    print("-", item.text.strip()) 
````

### 2.3 CSS 선택자 사용하기

````python
# class가 content인 모든 요소 찾기
content_elements = soup.select('.content')
print("\ncontent 클래스 내용:")
for element in content_elements:
    print("-", element.text.strip())
````

````python
# id가 title인 요소 모두 찾기
title = soup.select('#title')
print("\n제목:", title[0].text if title else "없음")
````

````python
# container 클래스 안의 모든 자손 p 태그 찾기
container_p = soup.select('.container p')
print("\ncontainer 안의 p 태그들:")
for p in container_p:
    print("-", p.text)
````

### 2.4 속성 다루기

````python
# 링크의 href 속성 가져오기
link = soup.find('a')
print("\n링크 주소:", link['href'])  
print("링크 텍스트:", link.text) 
````

````python
# 안전하게 속성 가져오기
print("링크 주소 (get 사용):", link.get('href'))
print("없는 속성:", link.get('title', '기본값'))
````

### 2.5 요소 탐색하기

````python
# 1. 부모 요소 찾기
p_tag = soup.find('p')
parent_div = p_tag.parent
print("p 태그의 부모 class:", parent_div.get('class'))  
````

````python
# 2. 자식 요소 찾기
container = soup.find('div', class_='container')
print("\n컨테이너의 직계 자식들:")
for child in container.children:
    if child.name:  # 공백 텍스트 제외
        print("-", child.name)  # h1, p, p, ul, a 출력
````

````python
# 3. 형제 요소 탐색
first_p = soup.find('p')
print("\n첫 번째 p의 다음 형제:")
next_sibling = first_p.find_next_sibling()
print("다음 형제 텍스트:", next_sibling.text)  
````

````python
second_p = next_sibling
print("두 번째 p의 이전 형제:")
prev_sibling = second_p.find_previous_sibling()  
print("이전 형제 텍스트:", prev_sibling.text) 
````

````python
# 4. 모든 형제 요소 찾기
print("\nli 태그 이후의 모든 형제:")
li_tag = soup.find('li')
for sibling in li_tag.find_next_siblings():
    print("-", sibling.name, ":", sibling.text)
````

````python
# 5. 특정 요소의 모든 상위 요소 찾기
li_tag = soup.find('li')
print("\n첫 번째 li의 모든 상위 요소:")
for parent in li_tag.parents:
    if parent.name:
        print("-", parent.name)
````

### 3. 여러 속성 동시 검색

````python
from bs4 import BeautifulSoup

# 예시 HTML
html = """
<html>
    <body>
        <div class="container" id="main">메인 컨테이너</div>
        <div class="container" id="sub">서브 컨테이너</div>
        <div class="content" id="article">
            <p class="text" id="first">첫 번째 문단</p>
            <p class="text highlight" id="second">두 번째 문단</p>
        </div>
        <span class="text" data-type="info">추가 정보</span>
        <p class="text highlight" id="ss">세 번째 문단</p>
        <p id="third">세 번째 문단</p>
        <p class="text highlight" id="third">세 번째 문단</p>
        <p class="text highlight" id="third">세 번째 문단</p>
        <div class="container">푸터 컨테이너</div>
    </body>
</html>
"""

soup = BeautifulSoup(html, 'html.parser')
````

````python
# 1. 기본적인 다중 속성 검색
result1 = soup.find('p', attrs={'class': 'text', 'id': 'first'})
print("1. class='text'이고 id='first'인 p태그:")
print(f"- {result1.text}")  # 출력: 첫 번째 문단
````

### 3.1 css 선택자로 계층적 탐색 및 여러 속성 검색

````python
# 3-1. 직계 자식 선택자 (>) 사용
print("1. 직계 자식 선택자:")
elements1 = soup.select("div.content > p.text")
for elem in elements1:
   print(f"- {elem.text}")
````

````python
# 3-2. 자손 탐색
print("\n2. 자손 선택자:")
elements2 = soup.select("div.content p.text")
for elem in elements2:
   print(f"- {elem.text}")
````

````python
# 3-3. ID 선택자 사용
print("\n3. ID 선택자:")
elements4 = soup.select("#article p")
for elem in elements4:
   print(f"- {elem.text}")
````

````python
# 3-4 여러 클래스 선택
print("\n4. 여러 클래스 선택:")
elements3 = soup.select("p.text.highlight")
for elem in elements3:
   print(f"- {elem.text}")
````

````python
# 3-5. 속성 선택자 사용
print("\n5. 속성 선택자:")
elements5 = soup.select("[data-type='info']")
for elem in elements5:
   print(f"- {elem.text}")  # 추가 정보 출력
````

````python
# 3-6. 여러 선택자 조합 사용
print("\n6. 복합 선택자:")
elements6 = soup.select("div.content > p.text#second")
for elem in elements6:
   print(f"- {elem.text}")  # 두 번째 문단 출력
````

````python
# 3-7. 속성과 텍스트 값 가져오기
print("\n7. 속성과 텍스트 값:")
elements7 = soup.select("div.container")
for elem in elements7:
   print(f"- id: {elem.get('id')}, text: {elem.text}")
````

### 💡 요약: 언제 무엇을 사용할까?

**`find` / `find_all` 를 사용하세요:**
- 간단한 태그 검색이 필요할 때 (예: 모든 `<p>` 태그 찾기)
- 태그와 클래스를 함께 지정할 때 (예: `class='container'`인 `<div>`)
- CSS 선택자에 대해 잘 모르지만 요소 검색이 필요한 경우 

**`select` / `select_one` 을 사용하세요:**
- 복잡한 계층 구조를 검색할 때 (예: `div > ul > li`)
- CSS 선택자에 익숙하거나 웹 개발 경험이 있을 때
- 한 줄로 간결하게 작성하고 싶을 때
